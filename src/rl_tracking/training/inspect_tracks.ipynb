```python
import numpy as np 
import matplotlib.pyplot as plt 
import numpy as np 
file_path = '/Users/liv/trackML/train_1/' 

```

```python
from trackml.dataset import load_event

hits, cells, particles, truth = load_event('/Users/liv/trackML/train_1/event000001024')

hits = hits.merge(truth, on='hit_id', how='left')

hits = hits.merge(particles, on='particle_id', how='left')

hits['r'] = np.sqrt(hits.x**2 + hits.y**2)
```

```python
non_zero_hits = hits[hits['particle_id'] != 0]
non_zero_hits.groupby(['volume_id', 'layer_id', 'particle_id']).size()
```

```python
vol_12 = hits[hits['volume_id'] == 13]
vol_12.groupby(['layer_id', 'particle_id']).size()


vol_12[vol_12['layer_id']==2].r.unique()
```
